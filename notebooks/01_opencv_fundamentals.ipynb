```python
# Import requested libraries.
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load the image using OpenCV
image = cv2.imread("../data/raw/drone_sample.jpg")
image
```

```python
# Get the dimensions of the image (height, width, channels)
image.shape
```

```python
# Calculate total pixels
total_pixels = image.shape[0] * image.shape[1]
print(f"Total count of pixels in the image: {total_pixels}")
```

```python
# Split the image into its Blue, Green, and Red color channels
b,g,r = cv2.split(image)
```

```python
# Resize the UAV image to 640x640 pixels
resized_image = cv2.resize(image, (640,640))

# Chechk the dimensions of the resized image
resized_image.shape
```

```python
# Display the resized image
plt.imshow(resized_image)
plt.axis("off")
```

```python
# Crop a specific region of the UAV image containing pedestrians and vehicles
cropped_image = image[310:600, 450:900]

# Convert the cropped image from BGR to RGB
cropped_rgb = cv2.cvtColor(cropped_image, cv2.COLOR_BGR2RGB)

# Display the cropped image
plt.imshow(cropped_rgb)
plt.axis("off")

# Chech the dimensions of the cropped image
cropped_image.shape
```

```python
# Convert the original image to grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Display the grayscale image
plt.imshow(gray_image, cmap = "gray")
plt.axis("off")

# Check the dimensions of the grayscale image
gray_image.shape
```

```python
# Convert the original UAV image from BGR to HSV
hsv_image = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

# Extract the Hue channel from the HSV image
hue = hsv_image[:, :, 0]

# Display the Hue channel
plt.imshow(hue, cmap = "hsv")
plt.axis("off")

# Check the dimensions of the HSV image
hsv_image.shape
```

```python
# Apply Gaussian Blur to the grayscale UAV image using a 5x5 kernel
blurred_image = cv2.GaussianBlur(gray_image, (5,5), sigmaX = 0)

# Display the blurred image in grayscale
plt.imshow(blurred_image, cmap = "gray")
plt.axis("off")

# Check the dimensions of the blurred image
blurred_image.shape
```

```python
# Apply Canny Edge Detection to the blurred UAV image
edges = cv2.Canny(blurred_image, 50, 150)

# Display the detected edges in grayscale
plt.imshow(edges, cmap = "gray")
plt.axis("off")

# Check the dimensions of the edge map
edges.shape
```
